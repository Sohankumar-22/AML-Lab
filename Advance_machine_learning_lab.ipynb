```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
x=np.array([1,2,3],dtype=float)
y=np.array([2,3,4],dtype=float)
n=len(x)
print("input x:",x)
print("actual y:",y)
print("number of sample :",n)
```

```python
w=0.0
b=0.0
alpha=0.05
iterations=2
print("intial weight(w):",w)
print("intial bias(b):",b)
print("learning rate(alpha):",alpha)
```

```python
#y=wx+b
def predict(x,w,b):
  return w*x+b
y_pred=predict(x,w,b)
print("predicted y:",y_pred)
```

```python
#j=1/m(sumetion over(y))
def compute_cost(y,y_pred):
  return np.sum((y-y_pred)**2)/(len(y))
y_pred=predict(x,w,b)
cost=compute_cost(y,y_pred)
print("cost:",cost)
```

```python
# Full iterative gradient descent function
def gradient_descent_iterative(x, y, w_initial, b_initial, alpha, iterations):
  w = w_initial
  b = b_initial
  n = len(x)
  cost_history = []

  for i in range(iterations):
    # Make prediction with current w and b
    y_pred = predict(x, w, b)

    # Calculate the gradients (partial derivatives of the cost function)
    # For cost J = 1/n * sum((y - y_pred)^2),
    # dJ/dw = (2/n) * sum((y_pred - y) * x)
    # dJ/db = (2/n) * sum(y_pred - y)

    dw = (2/n) * np.sum((y_pred - y) * x)
    db = (2/n) * np.sum(y_pred - y)

    # Update parameters (move in the opposite direction of the gradient)
    w = w - alpha * dw
    b = b - alpha * db

    # Calculate and store the cost for monitoring
    cost = compute_cost(y, y_pred) # Recompute y_pred with updated w,b for cost calculation
    cost_history.append(cost)

    # Print progress
    print(f"Iteration {i+1}/{iterations}: Cost = {cost:.4f}, w = {w:.4f}, b = {b:.4f}")

  return w, b, cost_history
```

```python
# Reset initial parameters before running iterative gradient descent
w = 0.0
b = 0.0

# Run the iterative gradient descent
w_optimized, b_optimized, cost_history = gradient_descent_iterative(x, y, w, b, alpha, iterations)

print(f"\nFinal optimized weight (w): {w_optimized:.4f}")
print(f"Final optimized bias (b): {b_optimized:.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Data Initialization
x = np.array([1, 2, 3], dtype=float)
y = np.array([2, 3, 4], dtype=float)
n = len(x)

# Initial Parameters
w = 0.0
b = 0.0
alpha = 0.05
iterations = 2 # Reverted to 2 iterations as requested by the user

# Prediction Function
def predict(x, w, b):
  return w * x + b

# Cost Function
def compute_cost(y, y_pred):
  return np.sum((y - y_pred)**2) / (len(y))

# Gradient Descent Function
def gradient_descent_iterative(x, y, w_initial, b_initial, alpha, iterations):
  w = w_initial
  b = b_initial
  n = len(x)
  cost_history = []

  for i in range(iterations):
    y_pred = predict(x, w, b)

    dw = (2/n) * np.sum((y_pred - y) * x)
    db = (2/n) * np.sum(y_pred - y)

    w = w - alpha * dw
    b = b - alpha * db

    cost = compute_cost(y, y_pred)
    cost_history.append(cost)

    # This line prints the cost and parameters for each iteration
    print(f"Iteration {i+1}/{iterations}: Cost = {cost:.4f}, w = {w:.4f}, b = {b:.4f}")

  return w, b, cost_history

# Run gradient descent
w_optimized, b_optimized, cost_history = gradient_descent_iterative(x, y, w, b, alpha, iterations)

print(f"Initial x: {x}")
print(f"Actual y: {y}")
print(f"Initial w: {w}, b: {b}, alpha: {alpha}, iterations: {iterations}")
print(f"Final optimized w: {w_optimized:.4f}")
print(f"Final optimized b: {b_optimized:.4f}")

# Plot cost history
plt.figure(figsize=(6,2))
plt.plot(range(1, iterations + 1), cost_history, marker='o', markersize=4, linestyle='-')
plt.title('Cost Function Over Iterations')
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.grid(True)
plt.show()

# Plot training data and regression line
y_final_pred = predict(x, w_optimized, b_optimized)

plt.figure(figsize=(6, 2))
plt.scatter(x, y, color='blue', label='Actual Data')
plt.plot(x, y_final_pred, color='red', linestyle='-', label=f'Regression Line (y = {w_optimized:.2f}x + {b_optimized:.2f})')
plt.title('Linear Regression with Gradient Descent')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
x1=np.array([-1,-1,1,1],dtype=float)
x2=np.array([-1,1,-1,1])
y=np.array([-2.2,-2.8,1.8,2.2],dtype=float)
n=len(x1) # Corrected to len(x1) from len(x)
#initialize
w1=0.0
w2=0.03333
b=0.0
alpha=0.1
lam=0.5
iteration=2
def prediction(x1, x2, w1, w2, b):
  return w1*x1 + w2*x2 + b
```

```python
def compute_lasso_cost(x1, x2, y, w1, w2, b, lam, n):
    y_pred = prediction(x1, x2, w1, w2, b)
    mse_cost = np.sum((y_pred - y)**2) / n
    l1_regularization = lam * (np.abs(w1) + np.abs(w2))
    total_cost = mse_cost + l1_regularization
    return total_cost

# Initial predictions and error calculation
initial_y_pred = prediction(x1, x2, w1, w2, b)
initial_error = initial_y_pred - y
print(f"Initial Predictions (y_pred): {initial_y_pred}")
print(f"Initial Error (y_pred - y): {initial_error}")

# Initial cost calculation
initial_cost = compute_lasso_cost(x1, x2, y, w1, w2, b, lam, n)
print(f"Initial Lasso Cost: {initial_cost:.4f}")
```

```python
def lasso_gradient_descent(x1, x2, y, w1, w2, b, alpha, lam, iterations, n):
    cost_history = []
    w1_history = []
    w2_history = []
    b_history = []

    for i in range(iterations):
        y_pred = prediction(x1, x2, w1, w2, b)

        # Gradients for MSE part
        dw1_mse = (2/n) * np.sum((y_pred - y) * x1)
        dw2_mse = (2/n) * np.sum((y_pred - y) * x2)
        db_mse = (2/n) * np.sum(y_pred - y)

        # Gradients for L1 regularization (subgradient)
        dw1_l1 = lam * np.sign(w1)
        dw2_l1 = lam * np.sign(w2)

        # Update parameters
        w1 = w1 - alpha * (dw1_mse + dw1_l1)
        w2 = w2 - alpha * (dw2_mse + dw2_l1)
        b = b - alpha * db_mse

        # Store history
        current_cost = compute_lasso_cost(x1, x2, y, w1, w2, b, lam, n)
        cost_history.append(current_cost)
        w1_history.append(w1)
        w2_history.append(w2)
        b_history.append(b)

        if (i + 1) % 1 == 0:
            print(f"Iteration {i+1}/{iterations}: Cost = {current_cost:.4f}, w1 = {w1:.4f}, w2 = {w2:.4f}, b = {b:.4f}")

    return w1, w2, b, cost_history, w1_history, w2_history, b_history

# Run Lasso Gradient Descent
w1_final, w2_final, b_final, cost_hist, w1_hist, w2_hist, b_hist = lasso_gradient_descent(
    x1, x2, y, w1, w2, b, alpha, lam, iteration, n
)

print(f"\nFinal Optimized Weights and Bias:")
print(f"w1: {w1_final:.4f}")
print(f"w2: {w2_final:.4f}")
print(f"b: {b_final:.4f}")
```

```python
plt.figure(figsize=(6, 6))
plt.plot(range(len(cost_hist)), cost_hist, marker='o', linestyle='-')
plt.title('Lasso Regression Cost History')
plt.xlabel('Iteration')
plt.ylabel('Cost')
plt.grid(True)
plt.show()

plt.figure(figsize=(6, 6))
plt.plot(range(len(w1_hist)), w1_hist, label='w1', marker='.')
plt.plot(range(len(w2_hist)), w2_hist, label='w2', marker='.')
plt.plot(range(len(b_hist)), b_hist, label='b', marker='.')
plt.title('Lasso Regression Weight and Bias History')
plt.xlabel('Iteration')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

```python
rf=["high","high","high","low","low","low"]
soil=["loamy","sandy","clay","loamy","sandy","clay"]
target=["good","good","good","poor","poor","good"]
df=pd.DataFrame({"rf":rf,"soil":soil,"target":target})
df
```

```python
# Implement OneR classifier
def oner_classifier(df, target_column):
    target_values = df[target_column].unique()
    features = [col for col in df.columns if col != target_column]

    best_feature = None
    min_error = float('inf')
    best_rules = {}

    print(f"Target column: '{target_column}' with values: {list(target_values)}\n")

    for feature in features:
        print(f"Analyzing feature: '{feature}'")
        feature_values = df[feature].unique()
        rules = {}
        errors = 0

        for f_val in feature_values:
            # Subset of data for current feature value
            subset = df[df[feature] == f_val]

            # Count occurrences of each target class in the subset
            class_counts = subset[target_column].value_counts()

            # The most frequent class is the prediction for this rule
            if not class_counts.empty:
                predicted_class = class_counts.idxmax()
            else:
                predicted_class = None # Should not happen with well-formed data

            rules[f_val] = predicted_class

            # Calculate errors for this specific rule
            # Misclassified count: total in subset - count of predicted class
            misclassified_count = len(subset) - class_counts[predicted_class] if predicted_class else len(subset)
            errors += misclassified_count

            print(f"  If {feature} is '{f_val}', predict '{predicted_class}' (Errors: {misclassified_count})")

        error_rate = errors / len(df)
        print(f"Total errors for '{feature}': {errors}/{len(df)} = {error_rate:.2f} error rate\n")

        if error_rate < min_error:
            min_error = error_rate
            best_feature = feature
            best_rules = rules

    print("--------------------------------------------------")
    print(f"Best Feature: '{best_feature}' with Min Error Rate: {min_error:.2f}")
    print("Best Rules:")
    for f_val, pred_class in best_rules.items():
        print(f"  If {best_feature} is '{f_val}' then predict '{pred_class}'")

    return best_feature, best_rules, min_error

# Run the OneR classifier on your DataFrame
best_feature, best_rules, min_error = oner_classifier(df, 'target')
```
